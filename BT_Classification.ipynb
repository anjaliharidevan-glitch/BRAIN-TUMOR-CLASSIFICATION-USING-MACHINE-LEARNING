```python
import os
import cv2
import numpy as np

data_dir = "/content/Training"

classes = ['glioma', 'meningioma', 'pituitary', 'notumor']

images = []
labels = []

for label in classes:
    folder_path = os.path.join(data_dir, label)

    for img in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img)

        img_array = cv2.imread(img_path)

        if img_array is not None:   # IMPORTANT CHECK
            img_array = cv2.resize(img_array, (128,128))
            images.append(img_array)
            labels.append(label)

X = np.array(images)
y = np.array(labels)

print("Image shape:", X.shape)
print("Labels shape:", y.shape)
```

```python
dataset_path = "/content/Training"
```

```python
images = []
labels = []

for class_name in classes:
    class_path = os.path.join(dataset_path, class_name)

    for img in os.listdir(class_path):
        img_path = os.path.join(class_path, img)

        image = cv2.imread(img_path)
        image = cv2.resize(image, (128,128))

        images.append(image)
        labels.append(class_name)
```

```python
# 1. Encode FIRST
le = LabelEncoder()
y = le.fit_transform(y)

# 2. THEN split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
plt.figure(figsize=(10,6))

for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(X_train[i])
    plt.title(classes[y_train[i]])
    plt.axis("off")

plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
```

```python
model = Sequential()

# First Convolution Layer
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(128,128,3)))
model.add(MaxPooling2D(2,2))

# Second Convolution Layer
model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

# Third Convolution Layer
model.add(Conv2D(128, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

# Flatten Layer
model.add(Flatten())

# Dense Layers
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))

# Output Layer (4 classes)
model.add(Dense(4, activation='softmax'))
```

```python
img = cv2.imread("sample_mri.jpg")
img = cv2.resize(img, (128,128))
img = img / 255.0
img = np.reshape(img, (1,128,128,3))

prediction = model.predict(img)

classes = labels  # use same labels from training!

print("Predicted:", classes[np.argmax(prediction)])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    validation_data=(X_test, y_test),
    batch_size=32
)
```

```python
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Accuracy:", test_accuracy)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

cm = confusion_matrix(y_test, y_pred_classes)

sns.heatmap(cm, annot=True, cmap="Blues")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
